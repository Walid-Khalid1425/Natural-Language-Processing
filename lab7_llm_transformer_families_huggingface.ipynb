# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

## Task 4 — Decoder-only Model: text generation with sampling

*(This task heading is missing in the original notebook, which goes from Task 3 to Task 5. It was added here because the Optional Challenge refers to a "sampling cell" and to `set_seed(42)`.)*

Use DistilGPT-2 to generate text with **sampling** instead of greedy decoding. Fixing the seed with `set_seed(42)` makes the result reproducible.

````python
set_seed(42)

prompt = "Natural language processing is"

samples = generator(
    prompt,
    max_new_tokens=30,
    do_sample=True,        # sample the next token instead of always taking the most likely one
    temperature=0.8,
    top_k=50,
    num_return_sequences=3
)

for i, sample in enumerate(samples, start=1):
    print(f"Sample {i}: {sample['generated_text']}\n")
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**Answer:** A chatbot such as ChatGPT has to *generate* a reply one token at a time, so it needs a model that predicts the next token from everything written so far, which is exactly what a decoder-only model does. Because of its causal (left-to-right) attention, the same model can read the user's prompt and the conversation history and continue the text with the answer, so no separate encoder is needed. Decoder-only models are also simple to train on huge amounts of raw text (just predict the next word), and they scale very well, which gives them strong conversation and instruction-following abilities.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
import torch

torch.seed()   # re-seed with a random value, i.e. NOT a fixed seed like set_seed(42)

prompt = "Natural language processing is"

for run in range(1, 4):
    out = generator(prompt, max_new_tokens=30, do_sample=True, temperature=0.8, top_k=50)
    print(f"Run {run}: {out[0]['generated_text']}\n")
````

**Answer:** The outputs are different because with `do_sample=True` the model does not always choose the most likely next token. It computes a probability distribution over the vocabulary and **randomly samples** the next token from it (controlled by `temperature` and `top_k`). Every run draws different random numbers, so the text goes in a different direction. With `set_seed(42)` the random numbers are the same each time, so the output is reproducible, and with greedy decoding (`do_sample=False`) there is no randomness at all.
